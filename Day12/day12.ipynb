```python
import pandas as pd
import numpy as np
x = np.array([10,20,30,40])
y = np.array([15,25,35,45])
# cov = np.cov(x,y)
df = pd.DataFrame({"X":x,"Y":y})
cov = df.cov()
```

```python
correlation = (cov)/(np.std(x)*np.std(y))
correlation
```

```python
x = np.array([10,20,30,40])
y = np.array([-15,-25,-35,-45])
cov = np.cov(x,y)
cov
```

```python
import seaborn as sns
from sklearn.datasets import load_iris
import pandas as pd
```

```python
X,y = load_iris(as_frame=True,return_X_y=True)
X.corr()
# y.head()
```

```python
sns.heatmap(X.corr(),annot=True)
```

```python
x = np.array([10,20,30,-400])
y = np.array([15,25,35,45])
df = pd.DataFrame({"X":x,"Y":y})
print("Skewness of X columes",df['X'].skew())
print("Skewness of Y columes",df['Y'].skew())
```

```python
x = np.array([10,20,30,-400])
y = np.array([15,25,35,50])
df = pd.DataFrame({"X":x,"Y":y})
print("Skewness of X columes",df['X'].kurtosis())
print("Skewness of Y columes",df['Y'].kurtosis())
```
