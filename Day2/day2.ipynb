```python
class Point:
    def __init__(self,xx,yy):
        self.x = xx
        self.y = yy

    def __str__(self):
        return f'({self.x}, {self.y})'
    
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)
    
p1 = Point(10,20)
p2 = Point(1,2)
p3 = p1 + p2  #p1.__add__(p2)
print(p3)
```

```python
#WAP to create a class named Distance which has feet and inches as data member.
#Overload any tow operators.

class Distance:
    def __init__(self,feet,inches):
        self.feet = feet
        self.inches = inches

    def __add__(self,other):
        distanceInInches = (self.feet + other.feet)*12 + self.inches + other.inches
        return Distance(distanceInInches//12,distanceInInches%12)
    
    def __sub__(self,other):
        distanceFirstInInches = self.feet * 12 + self.inches
        distanceSecondInInches = other.feet * 12 + other.inches
        diff = distanceFirstInInches - distanceSecondInInches
        return Distance(diff//12,diff%12) if diff > 0 else Distance(-diff//12,-diff%12)
    
    def __str__(self):
        totalDistanceInInches = self.feet*12 + self.inches
        return f"{totalDistanceInInches//12} feet and {totalDistanceInInches%12} inches"



d1 = Distance(2,8)
d2 = Distance(3,6)
d3 = d1 + d2  # d1.__add__(d2)
d4 = d1 - d2  # d1.__sub__(d2)
print(f"Total distance after addtion: {d3}")
print(f"Total distance after subbtraction: {d4}")
```
