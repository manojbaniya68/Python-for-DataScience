```python
from collections import deque
queue = deque([1,5,7])
print(queue)
queue.append(100)
print(queue)
queue.appendleft(200)
print(queue)
queue.pop()
print(queue)
queue.popleft()
print(queue)
```

```python
while queue:
    val = queue.pop()
    print(val)
```

```python
def squares(x):
    print("Outside Loop...")
    for i in range(x):
        print(f"Inside the loop... in {i+1}th iteration")
        yield i*i

# for x in squares(5):
#     print(x)
generator = squares(100)
print(type(generator))
print(next(generator))
print(next(generator))
print(next(generator))
```

```python
def myDecorators(func):
    def wrapper():
        print("Before")
        func()
        print("After")
    return wrapper

```

```python
@myDecorators
def sayHello():
    print("Hello")

sayHello()
```

```python
import time
def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time
        result = func(*args,**kwargs)
        end = time.time
        # print(f"Time elapsed = {end-start} seconds")
        print("hello")
        return result
    return wrapper

```

```python
@timer
def fact(n):
    f = 1
    for i in range(1,n+1):
        f = f*i
    return f
fact(4)
```

```python
#Use case I
students = [("Dharan","Ram",25),("Biratnagar","Shyam",10),("Itahari","Hari",20)]
students_sorted = sorted(students, key= lambda x : x[1])
print(students_sorted)
```

```python
#Use Case II : filtering
nums = [1,2,3,4,2,4,5,6,3]
evens = list(filter(lambda x: x%2 == 0,nums))
print(evens)
```

```python
# Use Case III : Mapping
nums = [1,2,3,4,2,4,5,6,3]
squares = list(map(lambda x : x*x,nums))
print(squares)
```
