```python
#Simple Inheritence
#Base Class
class Animal:
    def __init__(self,name):
        self.name = name
    
    def speak(self):
        print(f"{self.name} is Speaking")

#Derived class

class Dog(Animal):
    def __init__(self,name,bread):
        super().__init__(name)
        self.bread = bread

    def speak(self):
        super().speak()
        print(f"{self.name} of {self.bread} is Barking")

d1 = Dog("Lucy","Shiba Inu")
d1.speak()
```

```python
#Multiple Inheritance
class A:
    def show(self):
        print("A")

class B:
    def show(self):
        print("B")

class C(A,B):
    pass

c1 = C()
c1.show()
```

```python
#Method Resolution Order
C.mro()
```

```python
class DataList:
    def __init__(self,aList):
        self.aList = aList

    def mean(self):
        return sum(self.aList)/len(self.aList)
    
d = DataList([1,2,3,4,5])
d.mean()
```

```python
while True:

    try:
        n1 = input("First number : ")
        n2 = input("Second number : ")
        print(n1)
        if(n1 == 'q' or n2 == 'q'):
            break
        result = int(n1)/int(n2)
        print(result)
    except ZeroDivisionError:
        print("Cannot divide by zero")
    except ValueError:
        print("Please enter valid number")
    finally:
        print("This always runs")
```
