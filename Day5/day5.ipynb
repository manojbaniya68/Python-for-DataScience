```python
import csv
data = [
    {'Name': 'Ram', 'Address': 'Biratnagar', 'Age': '22'},
    {'Name': 'Shyam', 'Address': 'Itahari', 'Age': '23'},
    {'Name': 'Hari', 'Address': 'Dharan', 'Age': '21'},
]

with open(file='newnewnew.csv',mode="w",newline="") as f:
    writer = csv.DictWriter(f,fieldnames=["Name","Address","Age"])
    writer.writeheader()
    writer.writerows(data)
```

```python
import pandas as pd

df = pd.read_csv("newnewnew.csv")
print(df)
print(type(df))
```
