```python
# openweathermap.org

import requests
from pprint import pprint
url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url)
# print(response)
# print(type(response))

if response.status_code == 200:
    data = response.json()
    pprint(data)
else:
    print("Error : ",response.status_code)
```
